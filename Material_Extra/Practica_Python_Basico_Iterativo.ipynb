## 1. Funciones y Entrada de Usuario

Una **función** es un bloque de código reutilizable que realiza una tarea específica. En Python, se definen con la palabra clave `def`.

La función `input()` nos permite pedirle al usuario que ingrese información por teclado, y la función `print()` sirve para mostrar resultados por pantalla.

### 📚 Ejemplo: Suma de dos números

```
def sumar(a, b):
    """Esta función suma dos números."""
    resultado = a + b
    print(f"El resultado de sumar {a} y {b} es: {resultado}")

# Llamando a la función con valores fijos
sumar(5, 3)
```

### ¿Qué es el bloque `if __name__ == "__main__":`?
En Python, este bloque se utiliza para definir el "punto de entrada" principal de un programa. Indica que el código que está debajo solo debe ejecutarse si el archivo se ejecuta directamente, y no si se importa como un módulo desde otro archivo. Es una buena práctica para organizar el código.

### Bucle `for` (Para iterar)
Un bucle `for` nos permite recorrer elementos uno por uno. Lo usaremos para recorrer a todas las personas de nuestra base de datos, calcular su IMC y guardar el resultado en una nueva lista (un vector).

### 📚 Ejemplo: Bucle for y listas

```
# Ejemplo de cómo usar un for para llenar una lista vacía
numeros = [1, 2, 3]
dobles = [] # Creamos una lista (vector) vacía

for n in numeros:
    doble_n = n * 2
    dobles.append(doble_n) # Agregamos el resultado a la lista

print("Lista original:", numeros)
print("Vector con los dobles:", dobles)
```
