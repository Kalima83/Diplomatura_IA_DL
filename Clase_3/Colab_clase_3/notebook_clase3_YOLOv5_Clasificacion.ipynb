## Registro y automatización con ClearML

[ClearML](https://docs.ultralytics.com/yolov5/tutorials/clearml-logging-integration) está completamente integrado en YOLOv5 para rastrear su experimentación, administrar versiones de conjuntos de datos e incluso ejecutar ejecuciones de entrenamiento de forma remota. Para habilitar ClearML (verifique las celdas de arriba):

- `pip install clearml`
- ejecute `clearml-init` para conectarse a un servidor ClearML (**implemente su propio [servidor de código abierto](https://github.com/clearml/clearml-server)** o use nuestro [servidor alojado gratuito](https://docs.ultralytics.com/yolov5/tutorials/clearml-logging-integration))

Obtendrá todas las excelentes funciones esperadas de un administrador de experimentos: actualizaciones en vivo, carga de modelos, comparación de experimentos, etc., pero ClearML también rastrea cambios no confirmados y paquetes instalados, por ejemplo. ¡Gracias a eso, las tareas de ClearML (que es como llamamos a los experimentos) también son reproducibles en diferentes máquinas! Con solo 1 línea adicional, podemos programar una tarea de entrenamiento YOLOv5 en una cola para que sea ejecutada por cualquier número de agentes (trabajadores) de ClearML.

Puede usar ClearML Data para crear versiones de su conjunto de datos y luego pasarlo a YOLOv5 simplemente usando su ID único. Esto le ayudará a realizar un seguimiento de sus datos sin agregar problemas adicionales. ¡Explore el [Tutorial de ClearML](https://docs.ultralytics.com/yolov5/tutorials/clearml-logging-integration) para obtener más detalles!
